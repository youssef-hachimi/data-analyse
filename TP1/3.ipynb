```python

import pandas as pd                 
import matplotlib.pyplot as plt     
from fanalysis.pca import PCA       
```

```python
# 2. Charger le fichier CSV dans un DataFrame
df = pd.read_csv("data.csv")


df.head()
```

```python
df.info()
```

```python
df.head()
df = df.set_index("Country")
```

```python
acp = PCA(std_unit=True, stats=True,
          row_labels=df.index, col_labels=df.columns)
```

```python
# 6. Lancer le calcul de l'ACP sur les données numériques
acp.fit(df.values)
```

```python

print(acp.eig_)
```

```python

acp.plot_eigenvalues()
plt.show()
```

```python
# 9. Pourcentage cumulé de variance expliquée par les axes
acp.plot_eigenvalues(type="cumulative")
plt.show()
```

```python

acp.correlation_circle(num_x_axis=1, num_y_axis=2 , figsize=(10,8))
```
