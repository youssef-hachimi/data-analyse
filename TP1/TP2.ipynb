```python
import pandas as pd
# ------- pour importer l'acp
from fanalysis.pca import PCA
# ------- pour importer l'acp
```

```python
df = pd.read_table("pca_data.txt",header=0,index_col=0)
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.values
```

```python
#ACP normee
acp = PCA(std_unit=True,stats=True,row_labels=df.index, col_labels=df.columns)
```

```python
acp.fit(df.values)
```

```python
acp.eig_
```

## analyse de donness

```python
acp.plot_eigenvalues()
```

```python
# axe = valeur propre
```

```python
acp.plot_eigenvalues(type="cumulative")
```

### Graphiques factorials

```python
## Cercle des correlations
acp.correlation_circle(num_x_axis=1, num_y_axis=2,figsize=(10,8))
```

```python
# Mapping des points lignes
acp.mapping_row(num_x_axis=1, num_y_axis=2,figsize=(10,8))
```
